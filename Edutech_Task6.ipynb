```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder

# 1. Load the exact file from your folder
try:
    df = pd.read_csv('telco_churn (2).csv') 
    print("SUCCESS: File load ho gayi hai!")

    # 2. Data Cleaning:
    df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

    # 3. Task 6 Step: KNN Imputer (Advanced Imputation)
    numeric_df = df.select_dtypes(include=[np.number])
    imputer = KNNImputer(n_neighbors=5)
    df[numeric_df.columns] = imputer.fit_transform(numeric_df)
    print("Step 1: KNN Imputer complete.")

    # 4. Task 6 Step: Outlier Treatment (IQR Method)
    Q1 = df['MonthlyCharges'].quantile(0.25)
    Q3 = df['MonthlyCharges'].quantile(0.75)
    IQR = Q3 - Q1
    df = df[~((df['MonthlyCharges'] < (Q1 - 1.5 * IQR)) | (df['MonthlyCharges'] > (Q3 + 1.5 * IQR)))]
    print("Step 2: Outliers removed.")

    # 5. Task 6 Step: Encoding & Scaling
    # Text categories ko numbers mein badlo
    le = LabelEncoder()
    for col in df.select_dtypes(include=['object']).columns:
        if df[col].nunique() <= 2:
            df[col] = le.fit_transform(df[col])
    
    # Numbers ko scale karo (Standardization)
    scaler = StandardScaler()
    df[numeric_df.columns] = scaler.fit_transform(df[numeric_df.columns])
    print("Step 3: Encoding and Scaling done.")

    # 6. Final Deliverable: Save the cleaned file
    df.to_csv('Final_Task6_Submission.csv', index=False)
    print("\n" + "="*40)
    print("Mubarak ho! 'Final_Task6_Submission.csv' ban gayi.")
    print("="*40)

except Exception as e:
    print(f"Error: {e}")
```
